```python
import acemd
from simulate_structures_functions import *
import matplotlib.pyplot as plt
basepath = os.getcwd()+'/'
input_dict = json_dict(basepath+"GPCRs/input_realm.json")
resultspath = basepath + 'simulation_output/'


for entry in input_dict:  
    combined_fsims = []
    try: 
        name = entry['name']
        pdbcode = entry['pdbcode']
        apo = entry['apo']
        ligand_ = entry['ligands']
        lig_info = entry['ligands'][0]
        print(lig_info['resname'])
            
        proddir='%sproduction/%s/*/' % (resultspath, name)
        if os.path.exists(os.getcwd()+'%s_%s.png'% (name,lig_info['resname'])):
            print('%s_%s.png'% (name,lig_info['resname']))
            continue
        print(proddir)
        sims = simlist(glob(proddir), glob(proddir +'structure.pdb'))
        for traj_entry in sims:
            #print(traj_entry)
            traj_entry.trajectory= glob(proddir +'output_wrapped.xtc')
        
        fsims = simfilter(sims, './filtered/%s/' % name, filtersel='not (water or resname CLA or resname SOD or resname POPC)')
        combined_fsims = simmerge(combined_fsims,fsims)

        metr = Metric(combined_fsims)
        metr.set(MetricDistance('protein and name CA', 'not protein', periodic='selections', metric='contacts'))
        data = metr.project()
        data.fstep = 0.2
        tica = TICA(data, 4, units='ns')
        dataTica = tica.project(3)
        dataBoot = dataTica.bootstrap(0.8)
        dataBoot.cluster(MiniBatchKMeans(n_clusters=100))
        model = Model(dataBoot)
        model.plotTimescales(maxlag=15, units="ns")
        model.markovModel(10, 5, units='ns')
        mols = model.getStates()
        try:
            Path('plots/%s/' % lig_info['resname']).mkdir()
        except:
            print("folder exist")
            continue
        for index,m in enumerate(mols):
            m.write('plots/%s/%s_state%s.pdb' % (lig_info['resname'],lig_info['resname'],index))
        model.plotFES(0, 1, temperature=310, states=False,save = 'plots/%s/%s_totalFES.png'%(lig_info['resname'],lig_info['resname']) )
        model.plotFES(0, 1, temperature=310, states=True,save = 'plots/%s/%s_totalFES_states.png'%(lig_info['resname'],lig_info['resname']) )
        kin = Kinetics(model, temperature=310)
        kin.plotFluxPathways(statetype='macro', mode='net_flux', fraction=1.0, plot=True, save='plots/%s/%s_kinetics.png'%(lig_info['resname'],lig_info['resname']))
    except Exception as e:
        print("model "+name+" could not be build because ",e)
        print(traceback.format_exc())

    
```
